```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the first dataset
file_path_1 = 'p&r_values_first_plot.xlsx'
r_values_1 = pd.read_excel(file_path_1, sheet_name='r', index_col=0)
p_values_1 = pd.read_excel(file_path_1, sheet_name='p', index_col=0)

# Load the second dataset
file_path_2 = 'p&r_values_second_plot.xlsx'
r_values_2 = pd.read_excel(file_path_2, sheet_name='r', index_col=0)
p_values_2 = pd.read_excel(file_path_2, sheet_name='p', index_col=0)

# Insert blank rows after the 5th and 9th y-axis elements
def insert_blank_rows(df, positions):
    for pos in sorted(positions, reverse=True):
        blank_row = pd.DataFrame([[np.nan] * df.shape[1]], columns=df.columns, index=[' '])
        df = pd.concat([df.iloc[:pos], blank_row, df.iloc[pos:]], axis=0)
    return df

positions = [4, 13, 22]  # Positions after which to insert blank rows

r_values_1 = insert_blank_rows(r_values_1, positions)
r_values_2 = insert_blank_rows(r_values_2, positions)
p_values_1 = insert_blank_rows(p_values_1, positions)
p_values_2 = insert_blank_rows(p_values_2, positions)

# Function to generate asterisks based on p-values
def significance_stars(p):
    if p <= 0.0001:
        return '* * * *'
    elif p <= 0.001:
        return '* * *'
    elif p <= 0.01:
        return '* *'
    elif p <= 0.05:
        return '*'
    else:
        return ''

# Apply the function to create DataFrames of asterisks
stars_1 = p_values_1.applymap(significance_stars)
stars_2 = p_values_2.applymap(significance_stars)

# Find the common range of r-values for both heatmaps
vmin = min(r_values_1.min().min(), r_values_2.min().min())
vmax = max(r_values_1.max().max(), r_values_2.max().max())

# Adjusted Plotting
fig, axes = plt.subplots(1, 2, figsize=(13, 24), dpi=300, gridspec_kw={'width_ratios': [.80, 1], 'wspace': 0.05})  # Adjusted wspace for equal widths

sns.set(font_scale=1.2)  # Adjusted font scale for better readability

# Define a more appealing color palette and set NaNs to white
cmap = sns.diverging_palette(240, 10, n=9, as_cmap=True)
cmap.set_bad(color='white')  # Set NaN cells to white

# Function to customize the grid lines
def customize_grid(ax, data):
    nrows, ncols = data.shape
    
    for y in range(nrows):
        for x in range(ncols):
            if pd.isna(data.iloc[y, x]):
                # Draw white grid lines around the NaN cell
                ax.vlines(x=x, ymin=y, ymax=y+1, color='white', lw=1)
                ax.vlines(x=x+1, ymin=y, ymax=y+1, color='white', lw=1)
            else:
                # Draw black grid lines around non-NaN cells
                ax.vlines(x=x, ymin=y, ymax=y+1, color='grey', lw=1)
                ax.vlines(x=x+1, ymin=y, ymax=y+1, color='grey', lw=1)
                ax.hlines(y=y, xmin=x, xmax=x+1, color='grey', lw=1)
                ax.hlines(y=y+1, xmin=x, xmax=x+1, color='grey', lw=1)

    return ax

# Create the first heatmap
sns.heatmap(r_values_1, annot=stars_1, fmt='', cmap=cmap, center=0,
            linewidths=0.5, linecolor='grey',
            annot_kws={'size': 30, 'weight': 'bold', 'color': 'black', 'va': 'top'},
            ax=axes[0], vmin=vmin, vmax=vmax, cbar=False, 
            mask=r_values_1.isna())  # Mask NaN cells for annotation

# Overlay the grid lines for NaN cells
customize_grid(axes[0], r_values_1)

axes[0].xaxis.set_ticks_position('top')
axes[0].xaxis.set_label_position('top')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=-45, ha='right')
axes[0].set_yticklabels(axes[0].get_yticklabels(), rotation=0)

# Create the second heatmap
sns.heatmap(r_values_2, annot=stars_2, fmt='', cmap=cmap, center=0,
            linewidths=0.5, linecolor='grey',
            annot_kws={'size': 30, 'weight': 'bold', 'color': 'black', 'va': 'top'},
            ax=axes[1], vmin=vmin, vmax=vmax, 
            cbar_kws={'label': 'Correlation Coefficient (r)', 'shrink':0.4}, 
            mask=r_values_2.isna())  # Mask NaN cells for annotation

# Overlay the grid lines for NaN cells
customize_grid(axes[1], r_values_2)

# Increase column label font size
axes[0].tick_params(bottom=False, labelsize=22)  # Increase font size for first heatmap column labels
axes[1].tick_params(bottom=False, labelsize=22)  # Increase font size for second heatmap column labels


axes[1].xaxis.set_ticks_position('top')
axes[1].xaxis.set_label_position('top')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=-45, ha='right')
axes[1].set_yticklabels([''] * len(axes[1].get_yticklabels()))  # Remove y-axis labels from the second heatmap

plt.tight_layout()  # Adjust layout to fit all elements
plt.savefig("heatmap300dpi.tiff", dpi=300, bbox_inches='tight')  # Adjust dpi for desired resolution

# Display the plot
plt.show()
```
